## Memorize efficient matching method

```python
import healpy as hp
import numpy as np
from collections import defaultdict

NSIDE = 2048

def radec_to_pix(ra, dec, nside=NSIDE):
    theta = np.deg2rad(90 - dec)
    phi   = np.deg2rad(ra)
    return hp.ang2pix(nside, theta, phi)

def neighbor_pixels(ra, dec, radius, nside=NSIDE):
    vec = hp.ang2vec(ra, dec, lonlat=True)

    return hp.query_disc(nside, vec, radius, inclusive=True)

def angsep(ra1, dec1, ra2, dec2):
    ra1  = np.deg2rad(ra1)
    dec1 = np.deg2rad(dec1)
    ra2  = np.deg2rad(ra2)
    dec2 = np.deg2rad(dec2)

    return np.arccos(
        np.sin(dec1)*np.sin(dec2) + 
        np.cos(dec1)*np.cos(dec2)*np.cos(ra1 - ra2)
    )   # returns radians
```

```python
pix_gal = radec_to_pix(SDSS_gal['ra'], SDSS_gal['dec'], NSIDE)
```

```python
gal_index = defaultdict(list)
for i, p in enumerate(pix_gal):
    gal_id = SDSS_gal[i]['objid']
    gal_index[p].append([i, gal_id])
```

```python
from astropy.cosmology import Planck18 as cosmo

R_match = 8.0  # Mpc

cluster_matches = []

for i in range(len(rm_clus)):
    ra_c  = rm_clus[i]['RA_x']
    dec_c = rm_clus[i]['DEC_x']
    z_c   = rm_clus[i]['Z_SPEC_x']
    id_c = rm_clus[i]['ID']

    # projected comoving distance threshold
    D_M   = cosmo.comoving_transverse_distance(z_c).value  # Mpc
    theta_max_rad = R_match / D_M                          # radians

    # neighbor pixels
    neigh_pix = neighbor_pixels(ra_c, dec_c, theta_max_rad, NSIDE)

    # gather candidate galaxy indices

    cand = []
    for p in neigh_pix:
        cand.extend(gal_index.get(p, []))
    if not cand:
        continue

    cand = np.array(cand, dtype=np.int64)
    cand_ind = cand[:,0]
    cand_ID = cand[:,1]


    # angular separation
    sep = angsep(ra_c, dec_c, SDSS_gal[cand_ind]['ra'], SDSS_gal[cand_ind]['dec'])   # radians

    # compute projected comoving separation R_perp = D_M * theta
    Rperp = D_M * sep                                        # Mpc

    # mask for R_perp < 8 Mpc
    keep = Rperp < R_match

    # store matches
    for g in cand[keep]:
        ind_g =  g[0]
        id_g = g[1] 
        cluster_matches.append([i,ind_g])

cluster_matches = np.array(cluster_matches)
```

```python
## Construct a match between cluster and galaxy
```

```python
cl_matched_ind = cluster_matches[:,0]
gal_matched_ind = cluster_matches[:,1]
```

```python
from astropy.table import Table, hstack

# Extract matched rows
clus_matched = rm_clus[cl_matched_ind]
gal_matched = SDSS_gal[gal_matched_ind]

# Rename galaxy columns to avoid name collisions
gal_matched  = gal_matched .copy()
for col in gal_matched.colnames:
    gal_matched .rename_column(col, f"gal_{col}")

# Combine horizontally
matched = hstack([clus_matched, gal_matched])

#matched.write("cluster_galaxy_matched.fits", overwrite=True)
```

### Find galaxies within 8 Mpc region inside RM Center

```python
rm_z = rm_clus['Z_SPEC_x'] * cu.redshift
rm_d = rm_z.to(u.Mpc, cu.redshift_distance(Planck18, kind="comoving")) ##In units of Mpc/h
```

```python
rm_c = SkyCoord(ra=rm_clus['RA_x']*u.degree, dec=rm_clus['DEC_x']*u.degree, distance = rm_d)
gal_c = SkyCoord(ra=SDSS_gal['ra']*u.degree, dec=SDSS_gal['dec']*u.degree, distance = 1*u.Mpc)

## Search around each RM
max_sep = 1 * u.degree
idxc, idxcatalog, d2d, d3d = gal_c.search_around_sky(rm_c, max_sep)
np.all(d2d < max_sep)

rm_clus_matched = rm_clus[idxc]
SDSS_gal_matched = SDSS_gal[idxcatalog]
```

```python
def haversine(lon1, lat1, lon2, lat2, r=1):
    """
    Calculate the great circle distance in kilometers between two points (specified in decimal degrees)
    """
    # convert decimal degrees to radians 
    lon1 = np.radians(lon1)
    lat1 = np.radians(lat1)
    lon2 = np.radians(lon2)
    lat2 = np.radians(lat2)

    # haversine formula 
    dlon = lon2 - lon1 
    dlat = lat2 - lat1 
    a = np.sin(dlat/2)**2 + np.cos(lat1) * np.cos(lat2) * np.sin(dlon/2)**2
    c = 2 * np.arcsin(np.sqrt(a)) 
    return c*r
```

```python
#Compute distances
r_rm_gal = haversine(rm_clus_matched['RA_x'], rm_clus_matched['DEC_x'], SDSS_gal_matched['ra'], SDSS_gal_matched['dec'], r=1)
```
