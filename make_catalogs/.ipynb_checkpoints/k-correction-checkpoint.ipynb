## Import BGS (no k-corr)

```python
with open('bgs_clus_RM_gal_matched.pickle', 'rb') as handle:
    bgs_matched = pickle.load(handle)
```

```python
fcols = ['g','r','z','w1','w2']
for col in fcols:
    #bgs_matched['flux_'+col.lower()+'_dered'] = bgs_matched['FLUX_'+col]/data['MW_TRANSMISSION_'+col]
    bgs_matched['r_dered'] = 22.5 - 2.5*np.log10(bgs_matched['flux_r_dered'])
    bgs_matched['g_dered'] = 22.5 - 2.5*np.log10(bgs_matched['flux_g_dered'])
    bgs_matched['gmr'] = bgs_matched['g_dered']-bgs_matched['r_dered']
```

```python
cosmo = Planck18
h = 0.7
M_r = bgs_matched['r_dered']-cosmo.distmod(bgs_matched['Z_BGS']).value - 5*np.log10(h)
bgs_matched['M_r'] = M_r
```

## Import k-corrected 

```python
filename = '/global/cfs/cdirs/desi/survey/catalogs/DA2/LSS/loa-v1/LSScats/v1.1/PIP/BGS_ANY_clustering.dat.fits' 
kcorr_data = Table.read(filename)
```

### Choose only >0 fluxes

```python
cut = np.where((kcorr_data['flux_g_dered'] >0) & (kcorr_data['flux_r_dered'] >0) & (kcorr_data['flux_z_dered'] >0) & \
               (kcorr_data['flux_w1_dered'] >0) & (kcorr_data['flux_w2_dered'] >0))
kcorr_data = kcorr_data[cut]
```

```python
kcorr_data.columns
```

```python
fcols = ['g','r','z','w1','w2']
for col in fcols:
    #bgs_matched['flux_'+col.lower()+'_dered'] = bgs_matched['FLUX_'+col]/data['MW_TRANSMISSION_'+col]
    kcorr_data['r_dered'] = 22.5 - 2.5*np.log10(kcorr_data['flux_r_dered'])
    kcorr_data['g_dered'] = 22.5 - 2.5*np.log10(kcorr_data['flux_g_dered'])
    kcorr_data['gmr'] = kcorr_data['g_dered']-kcorr_data['r_dered']
```

```python
cosmo = Planck18
h = 0.7
M_r = kcorr_data['r_dered']-cosmo.distmod(kcorr_data['Z']).value - 5*np.log10(h)
kcorr_data['M_r'] = M_r
```

```python
## Changing names
kcorr_data.rename_columns(['r_dered', 'M_r'],['r_dered_kcorr', 'M_r_kcorr'])
```

```python
kcorr_data = kcorr_data['TARGETID','Z', 'r_dered_kcorr', 'M_r_kcorr']
```

## Match the k-corrected version catalog with the Y1 bgs_matched catalog

```python
kcorr_z = kcorr_data['Z'] * cu.redshift
kcorr_d = kcorr_z.to(u.Mpc, cu.redshift_distance(Planck18, kind="comoving")) ##In units of Mpc/h
```

```python
bgs_z = bgs_matched['Z_BGS'] * cu.redshift
bgs_d = bgs_z.to(u.Mpc, cu.redshift_distance(Planck18, kind="comoving")) #In units of Mpc/h
## Masking the bad redshifts
bgs_d[np.where(bgs_d < 0)] = 0 
```

```python
kcorr_c = SkyCoord(ra=kcorr_data['RA']*u.degree, dec=kcorr_data['DEC']*u.degree, distance = kcorr_d)
bgs_c = SkyCoord(ra=bgs_matched['RA_BGS']*u.degree, dec=bgs_matched['DEC_BGS']*u.degree, distance = bgs_d)

## Search around each RM
max_sep = 0.1 * u.arcsec
idxc, idxcatalog, d2d, d3d = bgs_c.search_around_sky(kcorr_c, max_sep)
np.all(d2d < max_sep)

kcorr_matched = kcorr_data[idxc]
bgs_kcorr_matched = bgs_matched[idxcatalog]
```
