```python
import numpy as np
import matplotlib.pyplot as plt

lambda_pivot = 60.0
xline = np.logspace(np.log10(20), np.log10(200), 300)

def pivoted_powerlaw(x, slope, pivot=lambda_pivot):
    """
    Relation normalized to pass through (pivot, pivot).
    Only the slope matters.
    """
    return pivot * (x / pivot) ** slope


def posterior_band_from_slopes(slopes, xline, pivot=lambda_pivot):
    """
    Convert posterior slope samples into median and 16-84% bands.
    """
    slopes = np.asarray(slopes, dtype=float)
    slopes = slopes[np.isfinite(slopes)]

    y_samples = np.asarray([
        pivoted_powerlaw(xline, slope, pivot=pivot)
        for slope in slopes
    ])

    y16, y50, y84 = np.nanpercentile(y_samples, [16, 50, 84], axis=0)
    slope16, slope50, slope84 = np.nanpercentile(slopes, [16, 50, 84])

    return {
        "y16": y16,
        "y50": y50,
        "y84": y84,
        "slope16": slope16,
        "slope50": slope50,
        "slope84": slope84,
    }


def plot_slope_posterior(ax, xline, slopes, label, color, ls="-", alpha=0.18):
    band = posterior_band_from_slopes(slopes, xline)

    ax.fill_between(
        xline,
        band["y16"],
        band["y84"],
        color=color,
        alpha=alpha,
        linewidth=0,
    )

    ax.plot(
        xline,
        band["y50"],
        color=color,
        linestyle=ls,
        lw=2,
        label=(
            label
            + rf", $m={band['slope50']:.2f}^{{+{band['slope84']-band['slope50']:.2f}}}_{{-{band['slope50']-band['slope16']:.2f}}}$"
        ),
    )

    return band
```

```python
import numpy as np
import matplotlib.pyplot as plt

lambda_pivot = 60.0
xline = np.logspace(np.log10(20), np.log10(200), 300)


def pivoted_powerlaw(x, slope, pivot=60.0):
    """
    Pivot-normalized relation.

    All relations pass through (pivot, pivot), so only the slope matters.
    """
    return pivot * (x / pivot) ** slope


def band_from_lnl_scatter(x, slope, sigma_lnl, pivot=60.0):
    """
    Convert quoted ln-scatter into a log10-space band around the pivoted relation.

    This follows the same convention as your notebook:
        sigma_log10(lambda) = sigma_lnl / ln(10)
        sigma_log10(y) = slope * sigma_log10(lambda)
    """
    y_med = pivoted_powerlaw(x, slope, pivot=pivot)

    sigma_log10_lambda = sigma_lnl / np.log(10)
    sigma_log10_y = slope * sigma_log10_lambda

    log_y_med = np.log10(y_med)

    y_low = 10 ** (log_y_med - sigma_log10_y)
    y_high = 10 ** (log_y_med + sigma_log10_y)

    return y_med, y_low, y_high


def plot_relation_with_band(
    ax,
    x,
    slope,
    sigma_lnl,
    label,
    color,
    linestyle="-",
    pivot=60.0,
    alpha=0.18,
):
    y_med, y_low, y_high = band_from_lnl_scatter(
        x,
        slope=slope,
        sigma_lnl=sigma_lnl,
        pivot=pivot,
    )

    ax.fill_between(
        x,
        y_low,
        y_high,
        color=color,
        alpha=alpha,
        linewidth=0,
    )

    ax.plot(
        x,
        y_med,
        color=color,
        linestyle=linestyle,
        lw=2,
        label=label,
    )

    return y_med, y_low, y_high


# ------------------------------------------------------------
# Literature relations from your notebook
# ------------------------------------------------------------

# Chunhao To+21
a_T21 = 0.943
sigma_lnl_T21 = 0.299
slope_T21 = 1.0 / a_T21

# Giles+22
slope_Tx_G22 = 0.59
sigma_lnl_Tx_G22 = 0.33

slope_Lx_G22 = 1.61
sigma_lnl_Lx_G22 = 1.07


# ------------------------------------------------------------
# Plot
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(8, 6), constrained_layout=True)

plot_relation_with_band(
    ax,
    xline,
    slope=slope_T21,
    sigma_lnl=sigma_lnl_T21,
    label=rf"To+21, $m={slope_T21:.2f}$",
    color="tab:purple",
    linestyle="-",
    pivot=lambda_pivot,
    alpha=0.18,
)

plot_relation_with_band(
    ax,
    xline,
    slope=slope_Tx_G22,
    sigma_lnl=sigma_lnl_Tx_G22,
    label=rf"Giles+22 $T_X$, $m={slope_Tx_G22:.2f}$",
    color="tab:green",
    linestyle="--",
    pivot=lambda_pivot,
    alpha=0.18,
)

# 1:1 reference
ax.plot(
    xline,
    xline,
    color="black",
    linestyle=":",
    lw=1.5,
    label="slope = 1",
)

# Common pivot marker
ax.scatter(
    [lambda_pivot],
    [lambda_pivot],
    color="black",
    s=35,
    zorder=10,
    label=rf"common pivot $\lambda_{{RM}}={lambda_pivot:.0f}$",
)

ax.set_xscale("log")
ax.set_yscale("log")

ax.set_xlabel(r"$\lambda_{\rm RM}$")
ax.set_ylabel(r"Pivot-normalized relation")

ax.set_xlim(20, 200)
ax.set_ylim(20, 200)

ax.legend(frameon=False, fontsize=9)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.table import unique

# ------------------------------------------------------------
# Setup
# ------------------------------------------------------------

table = bgs_matched

redmapper_col = "LAMBDA"
id_col = "ID"
z_col = "Z_SPEC_x"

lambda_pivot = 60.0
log_pivot = np.log10(lambda_pivot)

z_bins = [[0.1, 0.2], [0.2, 0.3], [0.3, 0.4]]
rm_trend_bins = np.logspace(np.log10(20), np.log10(100), 9)

desi_colors = ["#7f0000", "#b35c00", "#08306b"]
desi_markers = ["o", "s", "^"]

xline = np.logspace(np.log10(20), np.log10(200), 400)

redmapper_values = np.asarray(table[redmapper_col], dtype=float)
z_values = np.asarray(table[z_col], dtype=float)


# ------------------------------------------------------------
# Literature relations, pivot-normalized to lambda_RM = 60
# ------------------------------------------------------------

def pivoted_powerlaw(x, slope, pivot=60.0):
    return pivot * (x / pivot) ** slope


def band_from_lnl_scatter(x, slope, sigma_lnl, pivot=60.0):
    y_med = pivoted_powerlaw(x, slope, pivot=pivot)

    sigma_log10_lambda = sigma_lnl / np.log(10)
    sigma_log10_y = slope * sigma_log10_lambda

    log_y_med = np.log10(y_med)

    y_low = 10 ** (log_y_med - sigma_log10_y)
    y_high = 10 ** (log_y_med + sigma_log10_y)

    return y_med, y_low, y_high


def plot_literature_relation(
    ax,
    x,
    slope,
    sigma_lnl,
    label,
    color,
    linestyle="-",
    pivot=60.0,
):
    y_med, y_low, y_high = band_from_lnl_scatter(
        x,
        slope=slope,
        sigma_lnl=sigma_lnl,
        pivot=pivot,
    )

    ax.fill_between(
        x,
        y_low,
        y_high,
        color=color,
        alpha=0.07,
        linewidth=0,
        zorder=1,
    )

    ax.plot(
        x,
        y_med,
        color=color,
        linestyle=linestyle,
        lw=1.6,
        alpha=0.42,
        label=label,
        zorder=2,
    )


# Literature values from your notebook
slope_T21 = 1.0 / 0.943
sigma_lnl_T21 = 0.299

slope_Tx_G22 = 0.59
sigma_lnl_Tx_G22 = 0.33

slope_Lx_G22 = 1.61
sigma_lnl_Lx_G22 = 1.07


# ------------------------------------------------------------
# DESI binned points using log-mean and 16-84 scatter
# ------------------------------------------------------------

def compute_desi_binned_relation_for_zbin(zlo, zhi):
    zmask = (z_values >= zlo) & (z_values < zhi)
    zsub = table[zmask]
    zsub_rm_values = np.asarray(zsub[redmapper_col], dtype=float)

    x_log = []
    y_log = []
    y_p16_log = []
    y_p84_log = []
    n_bin = []

    for lo, hi in zip(rm_trend_bins[:-1], rm_trend_bins[1:]):
        mask = (zsub_rm_values >= lo) & (zsub_rm_values < hi)
        sub = zsub[mask]

        if len(sub) == 0:
            continue

        id_ind, _, lambda_true_ind = calc_specRichness_individual(
            binBoundaries,
            binCent,
            sub,
        )

        id_ind = np.asarray(id_ind)
        lambda_true_ind = np.asarray(lambda_true_ind, dtype=float)

        finite = np.isfinite(lambda_true_ind) & (lambda_true_ind > 0)
        id_ind = id_ind[finite]
        lambda_true_ind = lambda_true_ind[finite]

        sub_unique = unique(sub, keys=id_col)
        id_to_lambda_rm = {
            row[id_col]: float(row[redmapper_col])
            for row in sub_unique
        }

        lambda_rm_ind = []
        lambda_spec_ind = []

        for cid, lam_spec in zip(id_ind, lambda_true_ind):
            if cid in id_to_lambda_rm:
                lam_rm = id_to_lambda_rm[cid]
                if (
                    np.isfinite(lam_rm)
                    and lam_rm > 0
                    and np.isfinite(lam_spec)
                    and lam_spec > 0
                ):
                    lambda_rm_ind.append(lam_rm)
                    lambda_spec_ind.append(lam_spec)

        lambda_rm_ind = np.asarray(lambda_rm_ind, dtype=float)
        lambda_spec_ind = np.asarray(lambda_spec_ind, dtype=float)

        if len(lambda_spec_ind) <= 2:
            continue

        log_rm = np.log10(lambda_rm_ind)
        log_spec = np.log10(lambda_spec_ind)

        x_log.append(np.nanmean(log_rm))
        y_log.append(np.nanmean(log_spec))
        y_p16_log.append(np.nanpercentile(log_spec, 16))
        y_p84_log.append(np.nanpercentile(log_spec, 84))
        n_bin.append(len(log_spec))

    return {
        "x_log": np.asarray(x_log, dtype=float),
        "y_log": np.asarray(y_log, dtype=float),
        "y_p16_log": np.asarray(y_p16_log, dtype=float),
        "y_p84_log": np.asarray(y_p84_log, dtype=float),
        "n": np.asarray(n_bin, dtype=int),
    }


def pivot_normalize_desi_relation(data):
    """
    Shift only the intercept so the DESI relation passes through
    (lambda_RM, lambda_spec) = (60, 60). This preserves the slope and scatter.
    """
    x = data["x_log"]
    y = data["y_log"]

    good = np.isfinite(x) & np.isfinite(y)

    if np.sum(good) < 2:
        return None

    slope, intercept = np.polyfit(x[good], y[good], deg=1)

    y_at_pivot = slope * log_pivot + intercept
    shift = log_pivot - y_at_pivot

    out = dict(data)
    out["slope"] = slope
    out["intercept_original"] = intercept
    out["pivot_shift"] = shift
    out["y_log_norm"] = data["y_log"] + shift
    out["y_p16_log_norm"] = data["y_p16_log"] + shift
    out["y_p84_log_norm"] = data["y_p84_log"] + shift

    return out


# ------------------------------------------------------------
# Plot
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(8.5, 6.5), constrained_layout=True)

# Literature in transparent background
plot_literature_relation(
    ax,
    xline,
    slope=slope_T21,
    sigma_lnl=sigma_lnl_T21,
    label=rf"To+21, $m={slope_T21:.2f}$",
    color="tab:purple",
    linestyle="-",
    pivot=lambda_pivot,
)

plot_literature_relation(
    ax,
    xline,
    slope=slope_Tx_G22,
    sigma_lnl=sigma_lnl_Tx_G22,
    label=rf"Giles+22 $T_X$, $m={slope_Tx_G22:.2f}$",
    color="tab:green",
    linestyle="--",
    pivot=lambda_pivot,
)

# DESI highlighted with 16-84 scatter
for (zlo, zhi), color, marker in zip(z_bins, desi_colors, desi_markers):
    data = compute_desi_binned_relation_for_zbin(zlo, zhi)
    data = pivot_normalize_desi_relation(data)

    if data is None:
        continue

    x = 10 ** data["x_log"]
    y = 10 ** data["y_log_norm"]
    y_low = 10 ** data["y_p16_log_norm"]
    y_high = 10 ** data["y_p84_log_norm"]

    good = (
        np.isfinite(x)
        & np.isfinite(y)
        & np.isfinite(y_low)
        & np.isfinite(y_high)
        & (x > 0)
        & (y > 0)
        & (y_low > 0)
        & (y_high > 0)
    )

    ax.errorbar(
        x[good],
        y[good],
        yerr=[
            y[good] - y_low[good],
            y_high[good] - y[good],
        ],
        marker=marker,
        linestyle="none",
        capsize=4,
        color=color,
        ecolor=color,
        elinewidth=1.8,
        markeredgecolor="black",
        markeredgewidth=0.6,
        ms=8,
        label=rf"DESI DR1 ${zlo}<z<{zhi}$, $m={data['slope']:.2f}$",
        zorder=10,
    )

    # Draw a slope-only DESI guide line through the common pivot.
    yline_desi = pivoted_powerlaw(xline, data["slope"], pivot=lambda_pivot)

    ax.plot(
        xline,
        yline_desi,
        color=color,
        lw=2.8,
        alpha=0.95,
        zorder=9,
    )

# Common pivot and slope=1 reference
ax.scatter(
    [lambda_pivot],
    [lambda_pivot],
    color="black",
    s=45,
    zorder=20,
    label=rf"common pivot $\lambda_{{RM}}={lambda_pivot:.0f}$",
)

ax.plot(
    xline,
    xline,
    color="black",
    linestyle=":",
    lw=1.5,
    alpha=0.85,
    label="slope = 1",
    zorder=3,
)

ax.set_xscale("log")
ax.set_yscale("log")

ax.set_xlabel(r"$\lambda_{\rm RM}$")
ax.set_ylabel(r"Pivot-normalized richness relation")

ax.set_xlim(20, 200)
ax.set_ylim(15, 250)

ax.legend(frameon=False, fontsize=8, ncol=1)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.table import unique

# ------------------------------------------------------------
# Setup
# ------------------------------------------------------------

table = bgs_matched

redmapper_col = "LAMBDA"
id_col = "ID"
z_col = "Z_SPEC_x"

lambda_pivot = 60.0
log_pivot = np.log10(lambda_pivot)

z_bins = [[0.1, 0.2], [0.2, 0.3], [0.3, 0.4]]
rm_trend_bins = np.logspace(np.log10(20), np.log10(100), 9)

desi_colors = ["#7f0000", "#b35c00", "#08306b"]
desi_markers = ["o", "s", "^"]

xline = np.logspace(np.log10(20), np.log10(100), 400)

redmapper_values = np.asarray(table[redmapper_col], dtype=float)
z_values = np.asarray(table[z_col], dtype=float)


# ------------------------------------------------------------
# Literature relations, pivot-normalized to lambda_RM = 60
# ------------------------------------------------------------

def pivoted_powerlaw(x, slope, pivot=60.0):
    return pivot * (x / pivot) ** slope


def band_from_lnl_scatter(x, slope, sigma_lnl, pivot=60.0):
    y_med = pivoted_powerlaw(x, slope, pivot=pivot)

    sigma_log10_lambda = sigma_lnl / np.log(10)
    sigma_log10_y = slope * sigma_log10_lambda

    log_y_med = np.log10(y_med)

    y_low = 10 ** (log_y_med - sigma_log10_y)
    y_high = 10 ** (log_y_med + sigma_log10_y)

    return y_med, y_low, y_high


def plot_literature_relation(
    ax,
    x,
    slope,
    sigma_lnl,
    label,
    color,
    linestyle="-",
    pivot=60.0,
):
    y_med, y_low, y_high = band_from_lnl_scatter(
        x,
        slope=slope,
        sigma_lnl=sigma_lnl,
        pivot=pivot,
    )

    ax.fill_between(
        x,
        y_low,
        y_high,
        color=color,
        alpha=0.06,
        linewidth=0,
        zorder=1,
    )

    ax.plot(
        x,
        y_med,
        color=color,
        linestyle=linestyle,
        lw=1.6,
        alpha=0.38,
        label=label,
        zorder=2,
    )


# Literature values from your notebook
slope_T21 = 1.0 / 0.943
sigma_lnl_T21 = 0.299

slope_Tx_G22 = 0.59
sigma_lnl_Tx_G22 = 0.33

slope_Lx_G22 = 1.61
sigma_lnl_Lx_G22 = 1.07


# ------------------------------------------------------------
# DESI binned points using log-mean, SEM, and 16-84 scatter bands
# ------------------------------------------------------------

def compute_desi_binned_relation_for_zbin(zlo, zhi):
    zmask = (z_values >= zlo) & (z_values < zhi)
    zsub = table[zmask]
    zsub_rm_values = np.asarray(zsub[redmapper_col], dtype=float)

    x_log = []
    y_log = []
    y_sem_log = []
    y_p16_log = []
    y_p84_log = []
    n_bin = []

    for lo, hi in zip(rm_trend_bins[:-1], rm_trend_bins[1:]):
        mask = (zsub_rm_values >= lo) & (zsub_rm_values < hi)
        sub = zsub[mask]

        if len(sub) == 0:
            continue

        id_ind, _, lambda_true_ind = calc_specRichness_individual(
            binBoundaries,
            binCent,
            sub,
        )

        id_ind = np.asarray(id_ind)
        lambda_true_ind = np.asarray(lambda_true_ind, dtype=float)

        finite = np.isfinite(lambda_true_ind) & (lambda_true_ind > 0)
        id_ind = id_ind[finite]
        lambda_true_ind = lambda_true_ind[finite]

        sub_unique = unique(sub, keys=id_col)
        id_to_lambda_rm = {
            row[id_col]: float(row[redmapper_col])
            for row in sub_unique
        }

        lambda_rm_ind = []
        lambda_spec_ind = []

        for cid, lam_spec in zip(id_ind, lambda_true_ind):
            if cid in id_to_lambda_rm:
                lam_rm = id_to_lambda_rm[cid]
                if (
                    np.isfinite(lam_rm)
                    and lam_rm > 0
                    and np.isfinite(lam_spec)
                    and lam_spec > 0
                ):
                    lambda_rm_ind.append(lam_rm)
                    lambda_spec_ind.append(lam_spec)

        lambda_rm_ind = np.asarray(lambda_rm_ind, dtype=float)
        lambda_spec_ind = np.asarray(lambda_spec_ind, dtype=float)

        if len(lambda_spec_ind) <= 2:
            continue

        log_rm = np.log10(lambda_rm_ind)
        log_spec = np.log10(lambda_spec_ind)

        x_log.append(np.nanmean(log_rm))
        y_log.append(np.nanmean(log_spec))
        y_sem_log.append(np.nanstd(log_spec, ddof=1) / np.sqrt(len(log_spec)))
        y_p16_log.append(np.nanpercentile(log_spec, 16))
        y_p84_log.append(np.nanpercentile(log_spec, 84))
        n_bin.append(len(log_spec))

    return {
        "x_log": np.asarray(x_log, dtype=float),
        "y_log": np.asarray(y_log, dtype=float),
        "y_sem_log": np.asarray(y_sem_log, dtype=float),
        "y_p16_log": np.asarray(y_p16_log, dtype=float),
        "y_p84_log": np.asarray(y_p84_log, dtype=float),
        "n": np.asarray(n_bin, dtype=int),
    }


def pivot_normalize_desi_relation(data):
    """
    Shift only the intercept so the DESI relation passes through
    (lambda_RM, lambda_spec) = (60, 60). This preserves slope and scatter.
    """
    x = data["x_log"]
    y = data["y_log"]

    good = np.isfinite(x) & np.isfinite(y)

    if np.sum(good) < 2:
        return None

    slope, intercept = np.polyfit(x[good], y[good], deg=1)

    y_at_pivot = slope * log_pivot + intercept
    shift = log_pivot - y_at_pivot

    out = dict(data)
    out["slope"] = slope
    out["intercept_original"] = intercept
    out["pivot_shift"] = shift

    out["y_log_norm"] = data["y_log"] + shift
    out["y_sem_log_norm"] = data["y_sem_log"]
    out["y_p16_log_norm"] = data["y_p16_log"] + shift
    out["y_p84_log_norm"] = data["y_p84_log"] + shift

    return out


# ------------------------------------------------------------
# Plot
# ------------------------------------------------------------

fig, ax = plt.subplots(figsize=(8.5, 6.5), constrained_layout=True)

# Literature: faint background bands
plot_literature_relation(
    ax,
    xline,
    slope=slope_T21,
    sigma_lnl=sigma_lnl_T21,
    label=rf"To+21, $m={slope_T21:.2f}$",
    color="tab:purple",
    linestyle="-",
    pivot=lambda_pivot,
)

plot_literature_relation(
    ax,
    xline,
    slope=slope_Tx_G22,
    sigma_lnl=sigma_lnl_Tx_G22,
    label=rf"Giles+22 $T_X$, $m={slope_Tx_G22:.2f}$",
    color="tab:green",
    linestyle="--",
    pivot=lambda_pivot,
)

plot_literature_relation(
    ax,
    xline,
    slope=slope_Lx_G22,
    sigma_lnl=sigma_lnl_Lx_G22,
    label=rf"Giles+22 $L_X$, $m={slope_Lx_G22:.2f}$",
    color="tab:red",
    linestyle="--",
    pivot=lambda_pivot,
)

# DESI highlighted: 16-84 scatter bands + SEM error bars
shift = 1.02
for (zlo, zhi), color, marker in zip(z_bins, desi_colors, desi_markers):
    data = compute_desi_binned_relation_for_zbin(zlo, zhi)
    data = pivot_normalize_desi_relation(data)

    if data is None:
        continue

    x = 10 ** data["x_log"]
    y = 10 ** data["y_log_norm"]

    y_sem_low = 10 ** (data["y_log_norm"] - data["y_sem_log_norm"])
    y_sem_high = 10 ** (data["y_log_norm"] + data["y_sem_log_norm"])

    y_p16 = 10 ** data["y_p16_log_norm"]
    y_p84 = 10 ** data["y_p84_log_norm"]

    good = (
        np.isfinite(x)
        & np.isfinite(y)
        & np.isfinite(y_sem_low)
        & np.isfinite(y_sem_high)
        & np.isfinite(y_p16)
        & np.isfinite(y_p84)
        & (x > 0)
        & (y > 0)
        & (y_sem_low > 0)
        & (y_sem_high > 0)
        & (y_p16 > 0)
        & (y_p84 > 0)
    )

    xg = x[good]
    yg = y[good]
    y_sem_low_g = y_sem_low[good]
    y_sem_high_g = y_sem_high[good]
    y_p16_g = y_p16[good]
    y_p84_g = y_p84[good]

    order = np.argsort(xg)
    xg = xg[order]
    yg = yg[order]
    y_sem_low_g = y_sem_low_g[order]
    y_sem_high_g = y_sem_high_g[order]
    y_p16_g = y_p16_g[order]
    y_p84_g = y_p84_g[order]

    ax.fill_between(
        xg,
        y_p16_g,
        y_p84_g,
        color=color,
        alpha=0.22,
        linewidth=0,
        label=rf"DESI DR1 16--84 scatter, ${zlo}<z<{zhi}$",
        zorder=7,
    )

    ax.errorbar(
        xg,
        yg,
        yerr=[
            yg - y_sem_low_g,
            y_sem_high_g - yg,
        ],
        marker=marker,
        linestyle="none",
        capsize=4,
        color=color,
        ecolor=color,
        elinewidth=1.8,
        markeredgecolor="black",
        markeredgewidth=0.6,
        ms=8,
        label=rf"DESI DR1 mean $\pm$ SEM, ${zlo}<z<{zhi}$, $m={data['slope']:.2f}$",
        zorder=10,
    )

    yline_desi = pivoted_powerlaw(xline, data["slope"], pivot=lambda_pivot)

    ax.plot(
        xline,
        yline_desi,
        color=color,
        lw=2.8,
        alpha=0.95,
        zorder=9,
    )

# Pivot and reference
ax.scatter(
    [lambda_pivot],
    [lambda_pivot],
    color="black",
    s=45,
    zorder=20,
    label=rf"common pivot $\lambda_{{RM}}={lambda_pivot:.0f}$",
)

ax.plot(
    xline,
    xline,
    color="black",
    linestyle=":",
    lw=1.5,
    alpha=0.85,
    label="slope = 1",
    zorder=3,
)

ax.set_xscale("log")
ax.set_yscale("log")

ax.set_xlabel(r"$\lambda_{\rm RM}$")
ax.set_ylabel(r"Pivot-normalized richness relation")

ax.set_xlim(20, 100)
ax.set_ylim(5, 300)

ax.legend(frameon=False, fontsize=7.5, ncol=1)
plt.show()
```
