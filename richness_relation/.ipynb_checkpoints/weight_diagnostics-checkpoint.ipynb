# Richness Weight Diagnostics

This notebook visualizes the three multiplicative weights used in the spectroscopic-richness calculation:

- `WEIGHT`: DESI fiber/incompleteness weight, here called the IID weight.
- `1/geoFrac`: geometric coverage correction.
- `LF_WEIGHT`: luminosity-function completeness correction.

The combined richness weight is

```python
TOTAL_WEIGHT = IID_WEIGHT * GEOMETRIC_WEIGHT * LF_WEIGHT
```

The notebook shows histograms, redshift trends, and the combined-weight distribution.

````python
from pathlib import Path
import pickle
import sys

import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
from astropy.table import Table
from scipy.stats import binned_statistic

REPO_ROOT = Path('/global/homes/z/zzhang13/DESI/Projection')
if not REPO_ROOT.exists():
    REPO_ROOT = Path.cwd()
    if REPO_ROOT.name in {'richness_relation', 'tools', 'make_catalogs', 'plotting'}:
        REPO_ROOT = REPO_ROOT.parent

if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))

mpl.rcParams.update({
    'font.size': 12,
    'axes.linewidth': 1.1,
    'xtick.direction': 'in',
    'ytick.direction': 'in',
    'xtick.top': True,
    'ytick.right': True,
    'legend.frameon': False,
})

CATALOG_PATH = REPO_ROOT / 'catalogs' / 'bgs_clus_RM_gal_matched_with_geoFrac_lfweight.pickle'
OUTPUT_DIR = REPO_ROOT / 'plots'
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print('Repo root:', REPO_ROOT)
print('Catalog:', CATALOG_PATH)
````

````python
def read_pickle_table(path):
    with Path(path).open('rb') as handle:
        obj = pickle.load(handle)
    if isinstance(obj, Table):
        return obj
    if obj.__class__.__module__.startswith('pandas'):
        return Table.from_pandas(obj)
    if hasattr(obj, 'to_pandas'):
        return Table.from_pandas(obj.to_pandas())
    return Table(obj)

bgs = read_pickle_table(CATALOG_PATH)
print(f'Rows: {len(bgs):,}')
print('Required columns present:')
for col in ['IID_WEIGHT', 'geoFrac', 'GEOMETRIC_WEIGHT', 'LF_WEIGHT', 'TOTAL_WEIGHT', 'Z_SPEC_x', 'Z_BGS', 'ID']:
    print(f'  {col}:', col in bgs.colnames)
````

````python
# Weight components.
z_cluster = np.asarray(bgs['Z_SPEC_x'], dtype=float)
z_gal = np.asarray(bgs['Z_BGS'], dtype=float)

weight_iid = np.asarray(bgs['IID_WEIGHT'], dtype=float)
geo_frac = np.asarray(bgs['geoFrac'], dtype=float)
LF_WEIGHT = np.asarray(bgs['LF_WEIGHT'], dtype=float)

with np.errstate(divide='ignore', invalid='ignore'):
    weight_geo = 1.0 / geo_frac
    weight_total = weight_iid * weight_geo * LF_WEIGHT

valid = (
    np.isfinite(z_cluster)
    & np.isfinite(weight_iid)
    & np.isfinite(weight_geo)
    & np.isfinite(LF_WEIGHT)
    & np.isfinite(weight_total)
    & (weight_iid > 0)
    & (weight_geo > 0)
    & (LF_WEIGHT > 0)
    & (weight_total > 0)
)

weights = {
    'IID / fiber weight': weight_iid,
    'geometric weight': weight_geo,
    'LF weight': LF_WEIGHT,
    'combined weight': weight_total,
}

summary = []
for name, arr in weights.items():
    x = arr[valid]
    summary.append({
        'weight': name,
        'N': len(x),
        'min': np.nanmin(x),
        'p16': np.nanpercentile(x, 16),
        'median': np.nanmedian(x),
        'p84': np.nanpercentile(x, 84),
        'p95': np.nanpercentile(x, 95),
        'p99': np.nanpercentile(x, 99),
        'max': np.nanmax(x),
    })
summary_df = pd.DataFrame(summary)
display(summary_df)
````

## Histograms of Weight Components

The histograms use logarithmic x-axis scaling because the combined weight can have a long tail. The geometric component is plotted as `1/geoFrac`, since this is the multiplicative correction used in the richness weight.

````python
fig, axes = plt.subplots(2, 2, figsize=(11, 8), constrained_layout=True)
axes = axes.ravel()
colors = ['tab:blue', 'tab:orange', 'tab:green', 'tab:red']

for ax, (name, arr), color in zip(axes, weights.items(), colors):
    x = arr[valid]
    lo, hi = np.nanpercentile(x, [0.5, 99.5])
    bins = np.logspace(np.log10(max(lo, 1e-4)), np.log10(hi), 50)
    ax.hist(x, bins=bins, histtype='stepfilled', alpha=0.55, color=color)
    ax.axvline(np.nanmedian(x), color='black', ls='--', lw=1.2, label='median')
    ax.set_xscale('log')
    ax.set_xlabel(name)
    ax.set_ylabel('number of galaxies')
    ax.legend(loc='upper right')

fig.savefig(OUTPUT_DIR / 'richness_weight_component_histograms.png', dpi=180)
plt.show()
````

## Redshift Dependence

For each weight, the points show the median in bins of cluster redshift. The shaded band shows the 16th--84th percentile range within each redshift bin.

````python
def binned_percentiles(x, y, bins):
    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)
    centers = 0.5 * (bins[:-1] + bins[1:])
    med = np.full(len(centers), np.nan)
    p16 = np.full(len(centers), np.nan)
    p84 = np.full(len(centers), np.nan)
    count = np.zeros(len(centers), dtype=int)
    for i, (lo, hi) in enumerate(zip(bins[:-1], bins[1:])):
        m = (x >= lo) & (x < hi) & np.isfinite(y)
        count[i] = np.count_nonzero(m)
        if count[i] == 0:
            continue
        med[i] = np.nanmedian(y[m])
        p16[i], p84[i] = np.nanpercentile(y[m], [16, 84])
    return centers, med, p16, p84, count

z_bins = np.linspace(0.1, 0.4, 16)
fig, axes = plt.subplots(2, 2, figsize=(11, 8), sharex=True, constrained_layout=True)
axes = axes.ravel()

for ax, (name, arr), color in zip(axes, weights.items(), colors):
    centers, med, p16, p84, count = binned_percentiles(z_cluster[valid], arr[valid], z_bins)
    good = np.isfinite(med)
    ax.plot(centers[good], med[good], marker='o', color=color, label='median')
    ax.fill_between(centers[good], p16[good], p84[good], color=color, alpha=0.2, label='16-84%')
    ax.set_yscale('log')
    ax.set_ylabel(name)
    ax.legend(loc='upper left')

for ax in axes[-2:]:
    ax.set_xlabel(r'$z_{\rm BCG}$')

fig.savefig(OUTPUT_DIR / 'richness_weight_redshift_dependence.png', dpi=180)
plt.show()
````

## Combined Weight Versus Redshift

This plot shows all galaxies as a low-alpha scatter and overlays the median and 16th--84th percentile range. This is useful for identifying whether a small number of large weights dominate high-redshift richnesses.

````python
centers, med, p16, p84, count = binned_percentiles(z_cluster[valid], weight_total[valid], z_bins)
good = np.isfinite(med)

fig, ax = plt.subplots(figsize=(7.2, 5.2))
ax.scatter(z_cluster[valid], weight_total[valid], s=2, alpha=0.08, color='black', rasterized=True)
ax.plot(centers[good], med[good], marker='o', color='tab:red', lw=2, label='median')
ax.fill_between(centers[good], p16[good], p84[good], color='tab:red', alpha=0.22, label='16-84%')
ax.set_yscale('log')
ax.set_xlabel(r'$z_{\rm BCG}$')
ax.set_ylabel(r'combined weight $= {\rm WEIGHT}\,\times\,1/{\rm geoFrac}\,\times\,w_{\rm LF}$')
ax.legend(loc='upper left')
fig.tight_layout()
fig.savefig(OUTPUT_DIR / 'richness_combined_weight_vs_redshift.png', dpi=180)
plt.show()
````

## Correlation Matrix of Log Weights

This quick diagnostic shows whether the three corrections are independent or whether the same galaxies tend to receive large corrections in multiple weight components.

````python
log_weight_df = pd.DataFrame({
    'log10 IID': np.log10(weight_iid[valid]),
    'log10 geometric': np.log10(weight_geo[valid]),
    'log10 LF': np.log10(LF_WEIGHT[valid]),
    'log10 combined': np.log10(weight_total[valid]),
    'z_BCG': z_cluster[valid],
})

display(log_weight_df.corr())

fig, ax = plt.subplots(figsize=(5.6, 4.8))
im = ax.imshow(log_weight_df.corr(), vmin=-1, vmax=1, cmap='coolwarm')
ax.set_xticks(np.arange(len(log_weight_df.columns)))
ax.set_yticks(np.arange(len(log_weight_df.columns)))
ax.set_xticklabels(log_weight_df.columns, rotation=45, ha='right')
ax.set_yticklabels(log_weight_df.columns)
fig.colorbar(im, ax=ax, label='Pearson correlation')
fig.tight_layout()
fig.savefig(OUTPUT_DIR / 'richness_weight_log_correlation_matrix.png', dpi=180)
plt.show()
````
