# Spectroscopic richness offsets and distributions
The archived `old/richness_residual_fit.ipynb` plots conditional richness distributions, not fitted residuals. Here we retain its histogram-panel style and provide both distributions and the explicit offset $d=\log_{10}(\lambda_{\rm spec}/\lambda_{\rm RM})$. This is relative to equality, not a fitted residual or a pure projection estimate. The sign is the reverse of the earlier conditional-richness log ratio. No Gaussian/lognormal fits are drawn.
Rows represent weighting stages, columns represent redMaPPer bins, and colors represent the three BCG redshift bins. All stages use the same positive cluster sample. Histograms share edges across stages/redshifts within an RM bin. Raw richness PDFs are per unit linear richness; offset PDFs are per dex.
Run `python richness_relation/prepare_richness_weighting.py` once first. Both notebooks read the same cached cluster table; neither recalculates richness.

```python
from pathlib import Path
import sys
import matplotlib.pyplot as plt
from IPython.display import display
REPO_ROOT = Path('/global/homes/z/zzhang13/DESI/Projection')
if not REPO_ROOT.exists():
    REPO_ROOT = Path.cwd().resolve()
    while REPO_ROOT != REPO_ROOT.parent and not (REPO_ROOT / 'richness_relation').is_dir():
        REPO_ROOT = REPO_ROOT.parent
if not (REPO_ROOT / 'richness_relation').is_dir():
    raise FileNotFoundError('Set REPO_ROOT to the repository directory')
if str(REPO_ROOT) not in sys.path:
    sys.path.insert(0, str(REPO_ROOT))
from richness_relation.prepare_richness_weighting import load_stages
from richness_relation.plot_spectroscopic_richness_residuals import plot_distributions
from richness_relation.weighting_plot_utils import save
DATA_PATH = REPO_ROOT / 'catalogs/richness_weighting/cluster_weighting_stages.ecsv'
OUTPUT_DIR = REPO_ROOT / 'plots/richness_weighting'
RM_BINS = [20, 30, 50, 100]
MIN_COUNT = 20
HIST_BINS = 18
clusters, audit = load_stages(DATA_PATH)
display(audit)
```

```python
fig = plot_distributions(clusters, rm_bins=RM_BINS, residual=True, min_count=MIN_COUNT, hist_bins=HIST_BINS)
save(fig, OUTPUT_DIR, 'spectroscopic_richness_offsets')
plt.show()
plt.close(fig)
```

```python
fig = plot_distributions(clusters, rm_bins=RM_BINS, residual=False, min_count=MIN_COUNT, hist_bins=HIST_BINS)
save(fig, OUTPUT_DIR, 'spectroscopic_richness_distributions')
plt.show()
plt.close(fig)
```
